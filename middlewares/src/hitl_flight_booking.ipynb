For the flight booking application, Confirm the itinerary before purchase. This require to implement HITL before each step that is flight selection, final booking confirmation as well as for the cancellation.

HITL Points

-   Confirm travel details
-   Select the flight
-   Seat & add-ons (Window or aisle seat, Extra baggage, Meal preference, Travel insurance)
-   Final booking approval
-   HITL during cancellation & rescheduling
